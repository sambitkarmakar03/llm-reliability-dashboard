```python
import json
import numpy as np
import joblib
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# 1. Load the exported JSON
with open('scoring_model.json', 'r') as f: # replace with your exact filename
    model_data = json.load(f)

# 2. Reconstruct the StandardScaler
scaler = StandardScaler()
scaler.mean_ = np.array(model_data['scaler_mean'])
scaler.scale_ = np.array(model_data['scaler_scale'])
scaler.var_ = scaler.scale_ ** 2

# 3. Reconstruct the Logistic Regression model
lr = LogisticRegression()
lr.classes_ = np.array([0, 1])
lr.coef_ = np.array([model_data['coef']])
lr.intercept_ = np.array([model_data['intercept']])

# 4. Reconstruct the Pipeline
reconstructed_pipeline = Pipeline([
    ('scaler', scaler),
    ('lr', lr)
])

# 5. Export to PKL
joblib.dump(reconstructed_pipeline, 'scoring_model.pkl')
print("Successfully converted JSON weights back to scoring_model.pkl!")
```
